## DBCV

```python
import numpy as np
from sklearn.metrics import pairwise_distances
from tqdm import tqdm

def calculate_density(points, epsilon):
    local_dist_matrix = pairwise_distances(points)
    return np.sum(local_dist_matrix < epsilon, axis=1) / len(points)


def calculate_dbcv(data, clusters, epsilon=0.5):
    cluster_labels = np.unique(clusters)
    densities = []
    
    print("Calculating intra-cluster density")
    for label in tqdm(cluster_labels):
        if label == -1:  # Skip noise
            continue
        cluster_points = data[clusters == label]
        density = calculate_density(cluster_points, epsilon)
        densities.append(np.mean(density))
    intra_cluster_density = np.mean(densities)
    print("Intra-cluster density:", intra_cluster_density)

    print("Calculating inter-cluster density")
    inter_densities = []
    for label in tqdm(cluster_labels):
        if label == -1:
            continue
        cluster_points = data[clusters == label]
        density = calculate_density(cluster_points, epsilon)
        inter_densities.append(np.mean(density))
    inter_cluster_density = np.mean(inter_densities)
    print("Inter-cluster density:", inter_cluster_density)

    print("Calculating DBCV score")
    dbcv_score = intra_cluster_density / inter_cluster_density
    return dbcv_score
```

```python
coords = gdf[['latitude', 'longitude']].values
cluster_labels = gdf['hdbscan_cluster'].values
score = calculate_dbcv(coords, cluster_labels)
print(f"✅ Final DBCV Score: {score:.4f}" if not np.isnan(score) else "⚠️ DBCV Score could not be computed.")
```

## Soft Output

```python
import hdbscan

data = gdf[['latitude', 'longitude']]

clusterer = hdbscan.HDBSCAN(min_cluster_size=10, metric='haversine')
clusterer.fit(data)

soft_clusters = clusterer.probabilities_
gdf['soft_cluster_prob'] = soft_clusters
```

```python
import matplotlib.pyplot as plt

prob_average = gdf['soft_cluster_prob'].mean()
print(f"Average Soft Clustering Probability: {prob_average:.4f}")

plt.hist(gdf['soft_cluster_prob'], bins=50)
plt.title('Soft Clustering Probabilities')
plt.xlabel('Probability')
plt.ylabel('Frequency')
plt.show()
```

```python
persistence = clusterer.cluster_persistence_
print("Cluster Persistence:", persistence)
```

## Silhouette Score

```python
from sklearn.metrics import silhouette_score
sil_score = silhouette_score(data, clusterer.labels_, metric='haversine')
print("Silhouette Score:", sil_score)
```

## Davies-Bouldin Index

```python
from sklearn.metrics import davies_bouldin_score
import numpy as np

db_score = davies_bouldin_score(data, clusterer.labels_)
print("Davies-Bouldin Index (Euclidean):", db_score)
```
